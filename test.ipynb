```python
import json


data = {
    'no' : 1,
    'name' : 'Runoob',
    'url' : 'https://www.runoob.com'
}

json_str = json.dumps(data)
print ("Python 原始数据：", repr(data))
print ("JSON 对象：", json_str)

```

```python
# 写入 JSON 数据
with open('data.json', 'w') as f:
    json.dump(data, f)
    print("数据写入成功")
    print(type(data))
```

```python
# 读取数据
with open('data.json', 'r',encoding='UTF-8') as f:
    data = json.load(f)
    print(data)
    print(type(data))
```
