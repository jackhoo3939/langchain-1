# 展示 prompt 和 prompt template 的差别

```python
prompt = """
请总结下面的文章：
Python is a popular programming language.

"""
```

```python
template = """ 请把下面的{language} 翻译成{target_languate}:
{text}
"""
```

```python
language = "英文999"
target_languate = "中文999"
text = "太伟大了！Python is a popular programming language."
```

```python
print(template.format(language=language, target_languate=target_languate, text=text))
```
